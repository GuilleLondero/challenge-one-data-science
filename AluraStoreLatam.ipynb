### Importación de datos



```python
import pandas as pd

url1 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda1 = pd.read_csv(url1)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda1.head()
```

#1. Análisis de facturación: aquí se evaluará cual es la tienda que más vende
  
  
   




1.1 Cálculo del ingreso total de cada tienda:


```python
facturacion_tienda1 = tienda1['Precio'].sum()
facturacion_tienda2 = tienda2['Precio'].sum()
facturacion_tienda3 = tienda3['Precio'].sum()
facturacion_tienda4 = tienda4['Precio'].sum()
```

1.2 Resumen de la facturación obtenida en cada tienda:

```python
print(f"Total facturación Tienda 1: $ {round(facturacion_tienda1,2)} ")
print(f"Total facturación Tienda 2: $ {round(facturacion_tienda2,2)} ")
print(f"Total facturación Tienda 3: $ {round(facturacion_tienda3,2)} ")
print(f"Total facturación Tienda 4: $ {round(facturacion_tienda4,2)} ")
```

# 2. Ventas por categoría: se evaluará cual es el producto que más se vende en cada tienda
  

2.1 Cálculo de la cantidad de productos vendidos por tienda:

```python
ventas_por_categoria_tienda1 = tienda1['Categoría del Producto'].value_counts()
ventas_por_categoria_tienda2 = tienda2['Categoría del Producto'].value_counts()
ventas_por_categoria_tienda3 = tienda3['Categoría del Producto'].value_counts()
ventas_por_categoria_tienda4 = tienda4['Categoría del Producto'].value_counts()
```

2.2 Muestro los resultados:

```python
print("📦 Ventas por categoría en Tienda 1:")
print(ventas_por_categoria_tienda1)
print("\n📦 Ventas por categoría en Tienda 2:")
print(ventas_por_categoria_tienda2)
print("\n📦 Ventas por categoría en Tienda 3:")
print(ventas_por_categoria_tienda3)
print("\n📦 Ventas por categoría en Tienda 4:")
print(ventas_por_categoria_tienda4)
```

# 3. Calificación promedio de la tienda: objetivo conocer la satisfacción del cliente con los productos comprados:


3.1 Cálculo promedio calificaciones por tienda:

```python
prom_calificacion_tienda1 = round(tienda1['Calificación'].mean(), 2)
prom_calificacion_tienda2 = round(tienda2['Calificación'].mean(), 2)
prom_calificacion_tienda3 = round(tienda3['Calificación'].mean(), 2)
prom_calificacion_tienda4 = round(tienda4['Calificación'].mean(), 2)
```

3.2 Muestro los resultados:

```python
print("⭐ Valoración media por tienda:")

print("Tienda 1:", prom_calificacion_tienda1)
print("Tienda 2:", prom_calificacion_tienda2)
print("Tienda 3:", prom_calificacion_tienda3)
print("Tienda 4:", prom_calificacion_tienda4)
```

# 4. Productos más y menos vendidos por tienda:

4.1 Defino una función reutilizable para realizar el cálculo:

```python
def mostrar_productos_extremos(datos_tienda, nombre_tienda, cantidad):
    """
    Muestra los productos más vendidos y menos vendidos de una tienda.

    Parámetros:
    - df: DataFrame que contiene los datos de ventas de la tienda.
    - nombre_tienda: Nombre de la tienda (string) que se usará para los títulos en pantalla.
    - cantidad: Número de productos a mostrar.

    La función imprime por pantalla:
    - Los 'cantidad' productos más vendidos en la tienda.
    - Los 'cantidad' productos menos vendidos en la tienda.
    """
    conteo = datos_tienda['Producto'].value_counts()
    print(f"\n🔝 Productos más vendido en {nombre_tienda}:")
    print(conteo.head(cantidad))  # El más vendido

    print(f"\n🔻 Productos menos vendido en {nombre_tienda}:")
    print(conteo.tail(cantidad))  # El menos vendido
```

4.2 Muestro los resultados llamando a la función para cada tienda:

```python
mostrar_productos_extremos(tienda1, 'Tienda 1', 10)
mostrar_productos_extremos(tienda2, 'Tienda 2', 10)
mostrar_productos_extremos(tienda3, 'Tienda 3', 10)
mostrar_productos_extremos(tienda4, 'Tienda 4', 10)
```

# 5. Envío promedio por tienda:

5.1 Cálculo costo promedio de envío por tienda:

```python
prom_envio_tienda1 = round(tienda1['Costo de envío'].mean(), 2)
prom_envio_tienda2 = round(tienda2['Costo de envío'].mean(), 2)
prom_envio_tienda3 = round(tienda3['Costo de envío'].mean(), 2)
prom_envio_tienda4 = round(tienda4['Costo de envío'].mean(), 2)
```

5.2 Muestro los resultados:

```python
print("Costo promedio de envío por tienda:")

print(f"\nTienda 1:  $ {prom_envio_tienda1}")
print(f"Tienda 2:  $ {prom_envio_tienda2}")
print(f"Tienda 3:  $ {prom_envio_tienda3}")
print(f"Tienda 4:  $ {prom_envio_tienda4}")
```

# 6. Generando gráficos:

6.1 Gráfico de la facturación total de cada tienda:

```python
import matplotlib.pyplot as plt

# Creo las listas, en funcion de lo que quiero graficar:
facturacion = [facturacion_tienda1, facturacion_tienda2, facturacion_tienda3, facturacion_tienda4]
tiendas = ['Tienda 1', 'Tienda 2','Tienda 3', 'Tienda 4']

# Creo gráfico de barras para comparar la facturación:
plt.figure(figsize=(8, 5))  # Tamaño del gráfico

colores = ['seagreen', 'orange', 'steelblue', 'crimson'] #Asigno colores a cada tienda

plt.bar(tiendas, facturacion, color=colores) # tiendas:es el eje X; facturacion es el eje Y; color=colores le da un color distinto a cada barra.

# Título y etiqueta ejes:
plt.title('Facturación Total por Tienda', fontsize=14)
plt.xlabel('Tiendas', fontsize=12)
plt.ylabel('Facturación en pesos', fontsize=12)

# Mostrar los valores de facturación encima de cada barra
for i, valor in enumerate(facturacion):
    plt.text(i, valor + (max(facturacion)*0.01), f'{valor:,.2f}', ha='center', fontsize=10)

plt.tight_layout() #ajusta los márgenes automáticamente para que no se corten las etiquetas.
plt.show()
```

6.2 Gráfico calificación promedio de la tienda:

```python
# Creo las listas, en funcion de lo que quiero graficar:
tiendas = ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4']
valoraciones = [3.98, 4.04, 4.05, 4.00]

# Creo gráfico de líneas para comparar la puntuacón:
plt.figure(figsize=(8, 5))
plt.plot(tiendas, valoraciones, marker='o', linestyle='-', color='mediumslateblue', linewidth=3)

# Agregar puntos destacados con valores:
for i, val in enumerate(valoraciones):
    plt.text(i, val + 0.01, f'{val:.2f}', ha='center', fontsize=12)

# Título y etiqueta ejes:
plt.title('Valoración Promedio de los Clientes por Tienda', fontsize=14)
plt.xlabel('Tiendas', fontsize=12)
plt.ylabel('Calificación Promedio (1 a 5)', fontsize=12)
plt.ylim(3.9, 4.1)  # Enfoca el gráfico en el rango relevante

plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
# Datos de cada tienda
categorias = ['Muebles', 'Electrónicos', 'Juguetes', 'Electrodomésticos',
              'Deportes y diversión', 'Instrumentos musicales', 'Libros', 'Artículos para el hogar']

ventas_por_categoria_tienda1 = [465, 448, 324, 312, 284, 182, 173, 171]
ventas_por_categoria_tienda2 = [442, 422, 313, 305, 275, 224, 197, 181]
ventas_por_categoria_tienda3 = [499, 451, 315, 278, 277, 177, 185, 177]
ventas_por_categoria_tienda4 = [480, 451, 338, 254, 277, 170, 187, 201]

ventas_tiendas = [ventas_por_categoria_tienda1, ventas_por_categoria_tienda2, ventas_por_categoria_tienda3, ventas_por_categoria_tienda4]
nombres_tiendas = ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4']

# Creo subplots (figura con 4 espacios (2 filas x 2 columnas):
fig, axs = plt.subplots(2, 2, figsize=(12, 10))

# Paleta de colores consistente:
colores = plt.cm.Set3.colors

# Genero los 4 gráficos circulares (juntos):
for i, ax in enumerate(axs.flat): #Bucle recorre los 4 espacios del gráfico y dibuja un gráfico de torta en cada uno.
    ax.pie(ventas_tiendas[i],
           labels=categorias,
           autopct='%1.1f%%',
           startangle=140,
           colors=colores,
           wedgeprops={'edgecolor': 'black'})
    ax.set_title(f'Distribución por Categoría - {nombres_tiendas[i]}', fontsize=11)
    ax.axis('equal')

# Ajusto espacio entre gráficos
plt.suptitle('Comparación de Ventas por Categoría en las 4 Tiendas', fontsize=16)
plt.tight_layout(pad=3)
plt.show()
```
